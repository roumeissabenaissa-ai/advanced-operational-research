```python
#la lecture des donnes
def lire_donner(fishier):
 items = []
 with open (fishier , "r") as f :
 #with open ("knapsack_test100 (1).txt" , "r") as f :
 #with open ("knapsack_test1000.txt" , "r") as f :
   lignes = [l.strip() for l in f.readlines() if l.strip()]
 capacity = int (lignes[-1])
 for ligne in lignes[:-1]:
  w , v = ligne.split (",")
  items.append((int(w) , int (v)))
 valeur = [v for _, v in items]
 poid = [w for w, _ in items]
 return valeur, poid,  capacity

#Knapsack Greedy Algo
def knapsack( valeur,poid , capacity ):
   rapport = []
   select = []
   limit = 0
   v=0
   w=0

   for i in range(len(poid)):

    div = valeur[i]/poid[i]
    rapport.append((div , valeur[i],poid[i]))

   rapport.sort(reverse=True)

   while capacity> (limit + rapport[0][2]) and rapport:
    x = rapport.pop(0)
    select.append(x)
    limit +=x[2]

   print("les objest sélectionnés sont ")
   for i in range(len(select)):
    print(select[i][1] , select[i][2])
    v+=select[i][1]
    w+=select[i][2]

   print("le maximum ( total )  profit obtunue est " + str(v))
   print("le total poid obtunu " + str(w))
   print("le nombre des objets sélectionnés est " + str(len(select)))

import time
start_time = time.time()
valeur,poid,capacity = lire_donner("knapsack_test1000.txt")
knapsack( valeur , poid , capacity )
end_time = time.time()
print("Temps d'exécution méthode gloutonne:", end_time - start_time, "s")

```
